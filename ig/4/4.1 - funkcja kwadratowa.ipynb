(1 pkt) Uzupełnij wyrażenia obliczające rozwiązania równania kwadratowego. Wypisz na standardowe wyjście wszystkie rozwiązania lub komunikat o jego braku. Rozwiązania niekoniecznie są liczbami całkowitymi, zaokrąglij je do 2 miejsc po przecinku.

```
Równanie 1x^2 + -4x + 3 = 0
Rozwiązania: x_1 = 1.0, x_ 2 = 3.0
```

```
Równanie 9x^2 + -12x + 4 = 0
Rozwiązanie: x_1 = x_ 2 = 0.67
```

```
Równanie -6x^2 + 3x + -1 = 0
Rozwiązania: brak
```

````python
import math

# ma działać dla dowolnych a, b, c 
#(int i float, wczytywane od użytkownika - input)
a=input("Podaj a :") # a 
if '.' in a :
    a=float(a)
else :
    a=int(a)
b=input("Podaj b :") # b
if '.' in b :
    b=float(b)
else :
    b=int(b)
c=input("Podaj c :") # c
if '.' in c :
    c=float(c)
else :
    c=int(c)

delta = (b*b) -(4*a*c)           # zastąp zero wzorem na deltę
if delta > 0 :
    x_1 = round(((-b) -(math.sqrt(delta)))/(2*a),2)            # zastąp zero wzorem na rozwiązanie nr 1
    x_2 = round(((-b) +(math.sqrt(delta)))/(2*a),2)            # zastąp zero wzorem na rozwiązanie nr 2
elif delta == 0 :
    x_1 = x_2 = round((-b)/(2*a),2)

# Wypisz równanie
print("Równanie: ",a,"x^2 + ",b,"x + ",c," = 0 ",sep='')

# Wypisz rozwiązania
if delta > 0 :
    print("Rozwiązania: ", "x_1 = ",x_1,"," ," x_2 = ", x_2,sep='')
elif delta == 0 :
    print("Rozwiązanie: ", "x_1 = x_2 = ", x_1,sep='')
else :
    print("Rozwiązania: ", "brak",sep='')
````
