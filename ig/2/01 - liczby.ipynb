Proszę stworzyć program, który wczyta liczbę całkowitą podaną przez użytkownika. Wypisze tę liczbę podwojoną, a następnie powtórzy (konkatenacja) ją 5 razy. Następnie policzy i wyświetli tę liczbę podzieloną przez 3 (float), oraz wyświetl resztę z dzielenia przez 7. 

 

```
Podaj liczbę całkowitą: 6.
Twoja liczba po podwojeniu to: 12.
Powtarzając Twoją liczbę 5 razy otrzymamy: 66666.
Podzielona przez 3: 2.0.
Reszta z dzielenia przez 7 to: 6. 
```

````python
l=input("Podaj liczbę całkowitą :")# zastąpić odpowiednią funkcją wczytującą dane od użytkownika

# Wypisze wprowadzoną liczbę
print("Twoja liczba to: ", l,'.',sep='')
# Wypisze podwojoną
print("Twoja liczba po podwojeniu to: ", int(l) * 2,'.',sep='')
# Wypisze wiele razy
print("Powtarzając Twoją liczbę 5 razy otrzymamy: " , 5*l,'.',sep='')
# Wypisze podzieloną przez 3
print("Podzielona przez 3: ", float(l)/3,'.',sep='')
# Wypisze resztę z dzielenia
print("Reszta z dzielenia przez 7 to: ", int(l)%7,'.',sep='')
````
