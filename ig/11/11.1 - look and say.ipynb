Proszę napisać funkcję **next_look_and_say(n)**, zwracającą dla danej liczby będącej wyrazem ciągu [look-and-say](http://en.wikipedia.org/wiki/Look-and-say_sequence) kolejny jego wyraz. Funkcja ma działać dla różnych wyrazów początkowych. Funkcja nie powinna wykorzystywać _list comprehension_. 

Przetestuj ją, wykorzystując funkcje z poprzednich zadań:

```
>>> g = sequence_gen(next_look_and_say, 1)
>>> next(g)
1
>>> next(g)
11
>>> next(g)
21
>>> next(g)
1211
>>> next(g)
111221
>>> next(g)
312211
>>> next(g)
13112221
>>> next(g)
1113213211
```

````python
def next_look_and_say(n):
    "zwraca dla danej liczby będącej wyrazem ciągu look-and-say kolejny jego wyraz"
    strn = str(n)
    result = ""
    i = 0
    l = len(strn)
    while i < l:
        repeat = 1
        while i + 1 < l and strn[i] == strn[i+1]:
            repeat += 1
            i += 1
        result += str(repeat) + strn[i]
        i += 1
    return result

def sequence_gen(next_function, *initial):
    """
    Generujące kolejne wyrazy dowolnego ciągu danego przy pomocy definicji rekurencyjnej, dane jest n pierwszych wyrazów ciągu i funkcja obliczająca kolejny wyraz na podstawie n poprzednich, 
    przyjmuje jako kolejne argumenty: funkcję obliczającą kolejny wyraz na podstawie n poprzednich, a następnie n początkowych wyrazów ciągu
    """
    terms = initial

    for term in terms:
        yield term

    while True:
        next=next_function(*terms)
        terms = (*terms[1:], next)
        yield next


g = sequence_gen(next_look_and_say, 1)
print(next(g))
print(next(g))
print(next(g))
print(next(g))
print(next(g))
print(next(g))
print(next(g))
print(next(g))
````
