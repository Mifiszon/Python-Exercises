Proszę wykonać transliterację piosenki znajdującej się w pliku ru.txt i wynik zapisać do transliteracja.txt.

```python
def transliter(input_file, output_file):
    """
    funkcja transliteruje piosenke w pliku ru.txt i zapisuje do kolejnego pliku
    """
    russian = "АаБбВвГгДдЕеЁёЖжЗзИиЙйКкЛлМмНнОоПпРрСсТтУуФфХхЦцЧчШшЩщЪъЫыЬьЭэЮюЯя’’"
    letter = "AaBbVvGgDdEeЁёŽžZzIiJjKkLlMmNnOoPpRrSsTtUuFfHhCcČčŠšŜŝ″″Yy′′ÈèÛûÂâ’’"
    
    #print(len(russian))
    #print(len(letter))

    with open(input_file, 'r') as input_file:
        song = input_file.read()

    text = ''

    for char in song:
        if char in russian:
            index = russian.index(char)
            text += letter[index]
        else:
            text += char

    with open(output_file, 'w') as output_file:
        output_file.write(text)

transliter('ru.txt', 'transliteracja.txt')
```
