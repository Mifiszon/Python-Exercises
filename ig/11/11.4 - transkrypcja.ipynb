Proszę wykonać transkrypcję piosenki znajdującej się w pliku ru.txt i wynik zapisać do transkrypcja.txt.

```python
def transscript(input_file, output_file):
    """
    funkcja wykonuje transkrypcje piosenki w pliku ru.txt i zapisuje do kolejnego pliku
    """
    russian = "АаБбВвГгДдЕеЁёЖжЗзИиЙйКкЛлМмНнОоПпРрСсТтУуФфХхЦцЧчШшЩщЪъЫыЬьЭэЮюЯя’’"
    letter = "AaBbVvGgDdEeЁёŽžZzIiJjKkLlMmNnOoPpRrSsTtUuFfHhCcČčŠšŜŝ″″Yy′′ÈèÛûÂâ’’"

    def exceptions(char, prev_char):
        """
        funkcja do obslugi wyjatkow
        """
        if char == "е":
            if prev_char in "жлцчшщ":
                return "e"
            elif prev_char in "ъь":
                return "je"
            else:
                return "ie"
        elif char == "ё":
            if prev_char in "жлчшщ":
                return "o"
            elif prev_char in "ъь":
                return "jo"
            else:
                return "io"
        elif char == "и":
            if prev_char == "ь":
                return "ji"
            elif prev_char in "жцш":
                return "y"
            else:
                return "i"
        elif char == "ю":
            if prev_char == "л":
                return "u"
            elif prev_char in "ъь":
                return "ju"
            else:
                return "iu"
        elif char == "я":
            if prev_char == "л":
                return "a"
            elif prev_char in "ъь":
                return "ja"
            else:
                return "ia"
        elif char == "ь":
            if prev_char in "жлшчщ":
                return ""
            return "´"
        elif char == "ъ":
            return ""
        return None
    

    with open(input_file, 'r') as input_file:
        song = input_file.read()

    text = ''
    prev_char = None

    for char in song:
        if char in russian:
            index = russian.index(char)
            exception = exceptions(char, prev_char)
            if exception:
                text += exception
            else:
                text += letter[index]
        else:
            text += char
        prev_char = char
        
    with open(output_file, 'w') as output_file:
        output_file.write(text)

transscript('ru.txt', 'transkrypcja.txt')
```
