Proszę wczytać liczbę (input), a następnie wypisać ją słownie pamiętając o odmianie i zasadach polszczyzny.  

*Do rzędu wielkości 10^63 (decyliard). 

```
Podaj liczbę: 5527
pięć tysięcy pięćset dwadzieścia siedem

Podaj liczbę: 3007319
trzy miliony siedem tysięcy trzysta dziewiętnaście
```

````python
def numwords(num):
    "zamienia liczbę na słowa"
    units = ["", "jeden", "dwa", "trzy", "cztery", "pięć", "sześć", "siedem", "osiem", "dziewięć"]
    decimalSmall = ["dziesięć", "jedenaście", "dwanaście", "trzynaście", "czternaście", "piętnaście", "szesnaście", "siedemnaście", "osiemnaście", "dziewiętnaście"]
    decimalLarge = ["", "dziesięć", "dwadzieścia", "trzydzieści", "czterdzieści", "pięćdziesiąt", "sześćdziesiąt", "siedemdziesiąt", "osiemdziesiąt", "dziewięćdziesiąt"]
    hundreds = ["", "sto", "dwieście", "trzysta", "czterysta", "pięćset", "sześcset", "siedemset", "osiemset", "dziewięćset"]
    thousands = ["", "tysiąc", "milion", "miliard", "bilion", "biliard", "trylion", "tryliard", "kwardylion", "kwadryliard", "kwintylion", "kwintyliard", "sekstylion", 
                 "sekstyliard", "septylion", "septyliard", "oktylion", "oktyliard", "nonylion", "nonyliard", "decylion", "decyliard"]

    def division(n):
        """
        konwertuje setki/dziesiątki/jedności (3) na słowa
        """
        words = []
        # 100
        words.append(hundreds[n // 100])
        n = n % 100
        # 10-19
        if 10 <= n < 20:
            words.append(decimalSmall[n - 10])
            return ' '.join(words)
        # 20-99
        elif n >= 20:
            words.append(decimalLarge[n // 10])
            n = n % 10
        # 1-9
        if n > 0:
            words.append(units[n])

        return ' '.join(words)

    if num == 0:
        return "zero"

    parts = []
    group_num = 0
    while num > 0:
        group = num % 1000
        if group > 0:
            if group_num == 1 and group == 1:  # "tysiąc" should be singular
                parts.append(division(group) + " tysiąc")
            else:
                parts.append(division(group) + " " + thousands[group_num])
        num = num // 1000
        group_num += 1

    return ' '.join(reversed(parts)).strip()

num = int(input("Podaj liczbę : "))
print(numwords(num))
````
