(1 pkt) Wypisz w kolejnych wierszach liczby od 2 do x. Określ czy jest pierwsza czy złożona (i z czego). 

 

```
2 liczba pierwsza
3 liczba pierwsza
4 liczba złożona: 2*2
5 liczba pierwsza
6 liczba złożona: 2*3
```

```
7 liczba pierwsza
```

```
8 liczba złożona: 2*2*2
```

````python
def czy_pierwsze(x):
    "Wypisuje liczby od 2 do x. Określa czy są pierwsze czy złożone (i z jakich pierwszych się składają)."
    #
    # do uzupełnienia
    #
    for num in range(2, x+1):
        if len([x for x in range(1, num+1) if num%x == 0]) == 2:
            print(num, 'liczba pierwsza')
        else:
            factors = []
            divisor = 2
            n = num
            while n > 1:
                while n % divisor == 0:
                    factors.append(divisor)
                    n = n / divisor
                divisor += 1

            Sfactors = '*'.join(str(factor) for factor in factors)
            print(num, 'liczba złożona: ', Sfactors)
  
    
czy_pierwsze(8)
````
