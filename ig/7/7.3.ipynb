Proszę napisać w 2 (dwie!) dowolne funkcje, wykorzystujące **sensownie** wyrażenia lambda - przypadek ma być taki, w którym użycie tej konstrukcji językowej skraca lub upraszcza program. 

Podczas oddawania zadań proszę (w komentarzu) uzasadnić użycie wyrażenia lambda w tym przypadku.

Jedno z wyrażeń lambda powinno wykorzystywać argumenty z wartościami domyślnymi.

Można posłużyć się poprzednimi laboratoriami, usprawniając którąś z funkcji występujących w wykonanych wcześniej zadaniach.



```python
def czy_parzyste():
    """
    Funkcja sprawdza czy podana liczba jest parzysta
    """
    num = int(input("Podaj liczbę: "))
    l = lambda num, div=2: num%div == 0
    if (l(num)):
        print("Parzysta")
    else:
        print("Nieparzysta")

#czy_parzyste()

def multiplier():
    """
    Funkcja mnoży 2 dowolne liczby, wczytane przez użytkownika, przez siebie
    """
    a = int(input("Liczba 1: "))
    b = int(input("Liczba 2: "))
    multi = lambda a,b: a*b
    return multi (a,b)

#multiplier()

```
