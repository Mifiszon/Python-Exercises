Proszę napisać funkcję **get_values(iterator, n)**, zwracającą listę _n_ kolejnych wartości pobranych z iteratora. Gdy wartości się skończą, należy odpowiednio skrócić zwracaną listę. 

Funkcja powinna działać, jak poniżej:

```
>>> i = iter(range(10))
>>> get_values(i, 3)
[0, 1, 2]
>>> get_values(i, 5)
[3, 4, 5, 6, 7]
>>> get_values(i, 4)
[8, 9]
>>> get_values(i, 4)
[]
```

````python
def get_values(iterator, n):
    """
    Funkcja zwraca listę n kolejnych wartości pobranych z iteratora, po skończeniu wartości lista zostaje odpowiednio skrócona
    """
    l=[]
    for i in range(n):
        value = next(iterator, None)
        if value is None:
            break
        l.append(value)
    return l


i = iter(range(10))
print(get_values(i, 3))
print(get_values(i, 5))
print(get_values(i, 4))
print(get_values(i, 4))
````
