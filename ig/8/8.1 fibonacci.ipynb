Proszę nsapiać funkcję **fib_gen()**, zwracającą kolejne wartości ciągu Fibonacciego przy pomocy wyrażenia **yield**. Funkcja powinna przechowywać co najwyżej kilka kolejnych wyrazów ciągu (nie należy tworzyć np. listy wszystkich wartości).

Funkcja powinna działać, jak poniżej:

```
>>> f = fib_gen()
>>> next(f)
0
>>> next(f)
1
>>> next(f)
1
>>> next(f)
2
>>> next(f)
3
>>> next(f)
5
>>> next(f)
8
>>> next(f)
13
(...)
```

````python
def fib_gen():
    """
    Funkcja generuje i zwraca kolejne wartości ciągu Fibonacciego
    """
    a = 0
    b = 1
    while True:
        yield a
        c = a
        a = b
        b = c + b

f = fib_gen()
print(next(f))
print(next(f))
print(next(f))
print(next(f))
print(next(f))
print(next(f))
print(next(f))
print(next(f))

````
