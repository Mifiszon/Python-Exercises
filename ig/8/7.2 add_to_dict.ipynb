````python
def add_to_dict(key, value, *args, **kwargs):
    if args:
        d = args[0]
    else:
        d = {}
    d[key]=value
    return d

print(add_to_dict(1,2))
print(add_to_dict('a', 'b'))
print(add_to_dict('c', 'd', {5: 6}))
d = {'x': 'y'}
print(add_to_dict(10, 20, d))
d
````
