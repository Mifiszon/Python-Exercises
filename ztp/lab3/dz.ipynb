```python
def read_sales(path):
    with open(path, encoding="utf-8") as file:
        next(file)
        for line in file:
            line = line.strip()
            
            if not line:
                continue
                
            coll = line.split(",")
            quantity = float(coll[3])
            price = float(coll[4])
            sale = quantity * price
            if sale > 1000:
                yield sale

sales_gen = (x for x in read_sales("sales.csv") if x > 1000)
sales_list = list(sales_gen)
avg = sum(sales_list) / len(sales_list)

print("Sprzedaże > 1000:", sales_list)
print("Średnia sprzedaż:", avg)
```
