```python
class Reversed:
    def __init__(self, value):
        self.value = value
        self.index = len(self.value)

    def __iter__(self):
        return self

    def __next__(self):
        if self.index != 0:
            self.index -= 1
            return self.value[self.index]
        else:
            raise StopIteration

for x in Reversed("Python"):
    print(x, end=" ")
```
