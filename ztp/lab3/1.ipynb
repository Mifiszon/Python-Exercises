```python
class EvenNumbers:
    def __init__(self, start, end):
        if (start % 2) == 0:
            self.current = start
        else:
            self.current = start + 1
        if (end % 2) == 0:
            self.end = end
        else:
            self.end = end - 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.end:
            value = self.current
            self.current += 2
            return value
        else:
            raise StopIteration

for x in EvenNumbers(2, 10):
    print(x) 

print("\n")

for x in EvenNumbers(1, 11):
    print(x) 
```
