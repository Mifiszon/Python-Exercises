Proszę utworzyć plik osoby.csv z kolumnami: Imię, Nazwisko, Wiek.

Program ma odczytać dane, wypisać wszystkie wiersze i obliczyć średni wiek osób.

```python
import csv

def convert_int(value):
    try:
        return int(value)
    except ValueError:
        return value
        
with open("osoby.csv", newline='', encoding="utf-8") as f:
    reader = csv.reader(f, delimiter=';')
    age = 0
    counter = 0
    for imie, nazwisko, wiek in reader:
        wiek = convert_int(wiek)
        print(f"{imie} - {nazwisko} - {wiek}")
        if isinstance(wiek, int):
            age += wiek
            counter +=1
    print(f"\nŚredni wiek osób: {age/counter}")
    
```
