Proszę napisać program, który otworzy plik `text.txt` i wypisze numerowaną listę wierszy:

```
1: To jest pierwszy wiersz
2: To jest drugi wiersz
```

````python
with open("text.txt", "r", encoding="utf-8") as file:
    num = 1
    for line in file:
        if not line.strip():
            pass
        else:
            print(f"{num}:{line}")
            num += 1
````
