Proszę napisać funkcję, która przyjmie nazwę pliku i słowo kluczowe,a następnie wyświetli wszystkie wiersze zawierające to słowo.

```python
def keyword(path, word):
    with open(path, "r", encoding="utf-8") as file:
        for line in file:
            if word in line:
                print(line)
            else:
                pass

keyword("tekst.txt", "tu")
    
```
