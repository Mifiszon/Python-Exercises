Jest liczbą różnic pomiędzy napisami (stringami) o tej samej długości. (W przypadku ciągu bitów XOR. Będziemy operować tylko na napisach).

```
print( hamming('pies', 'pies') )
#0

print( hamming('granat', 'granit') )
#1

print( hamming('kotka', 'kotek') )
#2
```

````python
def hamming(word1, word2):
    counter = 0
    for i in range(len(word1)):
        if word1[i] != word2[i]:
            counter += 1
    return f'#{counter}'

print( hamming('pies', 'pies') )
print( hamming('granat', 'granit') )
print( hamming('kotka', 'kotek') )
````
