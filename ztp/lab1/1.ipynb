```python
import csv

'''
Część 1 – Wczytanie i przygotowanie danych
'''
try:
    with open("oceny.csv", newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        dane = list(reader)
        for x in dane:
            for key, value in x.items():
                if value in ['2', '3', '4', '5']:
                    x[key] = int(value)
                elif value in ['', 'nzal']:
                    x[key] = 'nzal'
        '''
        Część 2 – Analizy indywidualne
        '''
        #średnia ocen
        for x in dane:
            total = 0
            counter = 0
            gpa = 0
            for subject, grade in x.items():
                if isinstance(grade, int):
                    total += grade
                    counter += 1
            if counter > 0:
                gpa = round(total/counter, 2)
            else:
                gpa = 0
            x['średnia']= gpa

        #informacja o liczbie zaliczonych przedmiotów
        for x in dane:
            pass_counter = 0
            for subject, grade in x.items():
                # pomijamy pola nienależące do przedmiotów
                if subject not in ['imie', 'nazwisko', 'grupa', 'średnia', 'zaliczone']:
                    if grade in [3, 4, 5]:
                        pass_counter += 1
            x['zaliczone'] = pass_counter

        #ranking
        ranking = sorted(dane, key=lambda x: x['średnia'], reverse=True)
        
        '''
        print('___Ranking studentów___')
        for i in range(len(ranking)):
            x = ranking[i]
            print(f"{x['imie']} {x['nazwisko']} – Średnia: {x['średnia']}, Zaliczone: {x['zaliczone']}")

        
        #lista studentów bez nzal
        print('\n')
        print('___Studenci bez żadnego nzal___')
        for x in dane:
            if 'nzal' not in x.values():
                print(f"{x['imie']} {x['nazwisko']}")
        '''
        
        '''
        Część 3 – Analizy grupowe
        '''
        #średnia dla grup a i b
        gpa_list_A = []
        gpa_list_B = []
        
        for x in dane:
            total = 0
            counter = 0
            for subject, grade in x.items():
                if isinstance(grade, int):
                    total += grade
                    counter += 1
            if counter > 0:
                gpa = round(total/counter, 2)
            else:
                gpa = 0
            
            if x['grupa'] == 'A':
                gpa_list_A.append(gpa)
            elif x['grupa'] == 'B':
                gpa_list_B.append(gpa)
                
        #print('\n')       
        #print('___Średnie w grupach___')
        #print('A:', gpa_list_A)
        #print('B:', gpa_list_B)

        #porównanie średniej
        srednia_A = round(sum(gpa_list_A)/len(gpa_list_A), 2) 
        srednia_B = round(sum(gpa_list_B)/len(gpa_list_B), 2)

        '''
        print('\n')
        print('___Porównanie średniej___')
        print(f"Średnia grupy A: {srednia_A}")
        print(f"Średnia grupy B: {srednia_B}")
        
        if srednia_A > srednia_B:
            print('Grupa A wypadła lepiej')
        elif srednia_B > srednia_A:
            print('Grupa B wypadła lepiej')
        else:
            print('Średnie równe')
        '''

        #procent z 1 nzal w grupie
        nzal_counter_A = 0
        nzal_counter_B = 0
        counter_A = 0
        counter_B = 0
        
        for x in dane:
            grupa = x['grupa']
            has_nzal = False
            for subject, grade in x.items():
                if subject not in ['imie', 'nazwisko', 'grupa', 'średnia', 'zaliczone']:
                    if grade == 'nzal':
                        has_nzal = True
                        break
            if grupa == 'A':
                counter_A += 1
                if has_nzal:
                    nzal_counter_A += 1
            elif grupa == 'B':
                counter_B += 1
                if has_nzal:
                    nzal_counter_B += 1
                
        if counter_A > 0:
            percent_A = round((nzal_counter_A/counter_A)*100, 2)
        else:
            percent_A = 0

        if counter_B > 0:
            percent_B = round((nzal_counter_B/counter_B)*100, 2)
        else:
            percent_B = 0
        
        #print('\n')       
        #print('___Procent uczniów z nzal w grupach___')
        #print(f"Procent studentów z co najmniej jednym nzal w grupie A: {percent_A}%")
        #print(f"Procent studentów z co najmniej jednym nzal w grupie B: {percent_B}%")
        #print('\n')
        
        '''
        Część 4 – Analizy przedmiotowe
        '''
        #średnia ocen z przedmiotu
        subjects = ['matematyka', 'python', 'angielski', 'algorytmy', 'bazy_danych', 'ai']
        subject_avg = {}

        for subject in subjects:
            total = 0
            counter = 0
            for x in dane:
                if isinstance(x[subject], int):
                    total += x[subject]
                    counter += 1
            if counter > 0:
                avg = round(total/counter, 2)
            else:
                avg = 0
            subject_avg[subject]=avg

        '''
        print('___Średnie dla przedmiotów___')
        for x in subject_avg:
            print(f"{x}: {subject_avg[x]}")
        print('\n')  
        '''

        #najczęściej oceniany na 5
        subject_5 = {}
        for subject in subjects:
            counter = 0
            for x in dane:
                if x[subject] == 5:
                    counter += 1
            subject_5[subject] = counter
        
        max_count = -1
        subject_most_5 = ""
        for subject, counter in subject_5.items():
            if counter > max_count:
                max_count = counter
                subject_most_5 = subject

        #print('___Przedmiot najczęściej zaliczany na 5___')
        #print(f"Przedmiot najczęściej zaliczany na 5: {subject_most_5}")
        #print('\n') 
        
        
        #przedmiot z największą liczbą nzal
        subject_nzal = {}
        for subject in subjects:
            counter = 0
            for x in dane:
                if x[subject] == 'nzal':
                    counter += 1
            subject_nzal[subject] = counter
        
        max_count = -1
        subject_most_nzal = ""
        for subject, counter in subject_nzal.items():
            if counter > max_count:
                max_count = counter
                subject_most_nzal = subject

        #print('___Przedmiot z największą liczbą nzal___')
        #print(f"Przedmiot z największą liczbą nzal: {subject_most_nzal}")
        #print('\n') 
        

        '''
        Część 5 – Raport
        '''
        #zapis rankingu
        with open('ranking_studentow.csv', 'w', newline='', encoding='utf-8') as f:
            writer = csv.writer(f)
            writer.writerow(['Imię', 'Nazwisko', 'Grupa', 'Średnia', 'Zaliczone'])
            for x in ranking:
                writer.writerow([x['imie'], x['nazwisko'], x['grupa'], x['średnia'], x['zaliczone']])

        #zapis sutdentów bez nzal
        with open('studenci_bez_nzal.csv', 'w', newline='', encoding='utf-8') as f:
            writer = csv.writer(f)
            writer.writerow(['Imię', 'Nazwisko'])
            for x in dane:
                has_nzal = False
                for subject, grade in x.items():
                    if subject not in ['imie', 'nazwisko', 'grupa', 'średnia', 'zaliczone']:
                        if grade == 'nzal':
                            has_nzal = True
                            break
                if not has_nzal:
                    writer.writerow([x['imie'], x['nazwisko']])
        
        #zapis średnich uczniów w grupach
        with open('srednie_w_grupach.csv', 'w', newline='', encoding='utf-8') as f:
            writer = csv.writer(f)
            writer.writerow(['Imię', 'Nazwisko', 'Grupa', 'Średnia'])
            for x in dane:
                if x['grupa'] == 'A' or x['grupa'] == 'B':
                    writer.writerow([x['imie'], x['nazwisko'], x['grupa'], x['średnia']])

        #zapis średnich grup
        with open('srednie_grup.csv', 'w', newline='', encoding='utf-8') as f:
            writer = csv.writer(f)
            writer.writerow(['Grupa', 'Średnia'])
            writer.writerow(['A', srednia_A])
            writer.writerow(['B', srednia_B])
        
        #zapis procent uczniów z nzal
        with open('procent_z_nzal.csv', 'w', newline='', encoding='utf-8') as f:
            writer = csv.writer(f)
            writer.writerow(['Grupa', 'Procent z nzal'])
            writer.writerow(['A', percent_A])
            writer.writerow(['B', percent_B])

        #zapis średnich przedmiotów
        with open('srednie_przedmiotow.csv', 'w', newline='', encoding='utf-8') as f:
            writer = csv.writer(f)
            writer.writerow(['Przedmiot', 'Średnia'])
            for subject in subjects:
                writer.writerow([subject, subject_avg[subject]])
        
        #zapis przedmiot najczesciej zaliczany na 5
        with open('przedmiot_najczesciej_5.csv', 'w', newline='', encoding='utf-8') as f:
            writer = csv.writer(f)
            writer.writerow(['Przedmiot'])
            writer.writerow([subject_most_5])
        
        #zapis przedmiotu z najwieksza liczba zal
        with open('przedmiot_najwiecej_nzal.csv', 'w', newline='', encoding='utf-8') as f:
            writer = csv.writer(f)
            writer.writerow(['Przedmiot'])
            writer.writerow([subject_most_nzal])

        #raport w konsoli
        print('___RAPORT___')
        
        #5 najlepszych studentów
        print('5 najlepszych studentów:')
        for i in range(5):
            x = ranking[i]
            print(f"{i+1}. {x['imie']} {x['nazwisko']} – Średnia: {x['średnia']}, Zaliczone: {x['zaliczone']}")
        
        #średnie dla grup
        print('\n') 
        print('Średnie ocen dla grup:')
        print(f"Grupa A: {srednia_A}")
        print(f"Grupa B: {srednia_B}")
        
        #najłatwiejszy i najtrudniejszy przedmiot
        max_avg = -1
        min_avg = 6
        najlatwiejszy = ''
        najtrudniejszy = ''
        
        for subject in subjects:
            avg = subject_avg[subject]
            if avg > max_avg:
                max_avg = avg
                najlatwiejszy = subject
            if avg < min_avg:
                min_avg = avg
                najtrudniejszy = subject
        
        print(f"\nNajłatwiejszy przedmiot: {najlatwiejszy}")
        print(f"Najtrudniejszy przedmiot: {najtrudniejszy}")

except FileNotFoundError:
    print("Błąd: nie znaleziono pliku z danymi.")
except Exception as e:
    print("Nieoczekiwany błąd podczas wczytywania danych:", e)
```
