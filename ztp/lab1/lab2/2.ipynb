```python
import random

class Employee:
    def __init__(self, name):
        self.name = name

    def work(self):
        print(f"Cześć, jestem {self.name}.")

class Developer(Employee):
    def work(self):
        print(f"Cześć, jestem {self.name}. Programuję aplikacje.")

class Designer(Employee):
    def work(self):
        print(f"Cześć, jestem {self.name}. Projektuję interfejsy i grafiki.")

class Manager(Employee):
    def work(self):
        print(f"Cześć, jestem {self.name}. Zarządzam zespołem i projektami.")

names = ["Kamil", "Kamila", "Krzysztof", "Krystyna", "Maciej", "Michalina", "Michał"]
roles = [Developer, Designer, Manager]

for i in range(10):
    role_class = random.choice(roles)
    name = random.choice(names)
    employee = role_class(name)
    employee.work()
```
