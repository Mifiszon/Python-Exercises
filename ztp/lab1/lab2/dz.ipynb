```python
import logging

class Logger:
    _instance = None

    def __new__(cls):
        if not cls._instance:
            cls._instance = super().__new__(cls)
            cls._instance.logger = logging.getLogger("RentLogger")
            cls._instance.logger.setLevel(logging.INFO)
            file_handler = logging.FileHandler("rent.log", encoding='utf-8')
            formatter = logging.Formatter('%(asctime)s - %(levelname)s - %(message)s')
            file_handler.setFormatter(formatter)
            cls._instance.logger.addHandler(file_handler)
        return cls._instance

    def log(self, message):
        self.logger.info(message)
        print(f"[LOG] {message}")

class Vehicle:
    def __init__(self, id_, name):
        self.id = id_
        self.name = name
        self.access = False

    def rent(self):
        if not self.access:
            self.access = True
            return True
        return False

    def return_(self):
        if self.access:
            self.access = False
            return True
        return False

    def __str__(self):
        if self.access:
            state = "wypożyczony"
        else:
            state = "dostępny"
        return f"{self.id}: {self.name} ({state})"

class Car(Vehicle):
    pass

class Bike(Vehicle):
    pass

class Truck(Vehicle):
    pass

class Rental:
    def __init__(self, vehicles):
        self.vehicles = vehicles
        self.logger = Logger()

    def list_all(self):
        print("\n")
        print("Lista pojazdów:")
        for x in self.vehicles:
            print(x)
        print("\n")

    def rent(self, vehicle_id):
        vehicle = self.find_vehicle(vehicle_id)
        if vehicle:
            if vehicle.rent():
                self.logger.log(f"Wypożyczono pojazd {vehicle.name}")
            else:
                self.logger.log(f"Pojazd {vehicle.name} niedostępny")

    def return_(self, vehicle_id):
        vehicle = self.find_vehicle(vehicle_id)
        if vehicle:
            if vehicle.return_():
                self.logger.log(f"Zwrócono pojazd {vehicle.name}")
            else:
                self.logger.log(f"Pojazd {vehicle.name} nie został jeszcze wypożyczony")

    def find_vehicle(self, vehicle_id):
        for x in self.vehicles:
            if x.id == vehicle_id:
                return x
        return None

rental = Rental([Car(1, "Mazda"), Bike(2, "Trek"), Truck(3, "Volvo")])
rental.list_all()
rental.rent(1)
rental.rent(2)
rental.return_(1)
rental.list_all()
```
