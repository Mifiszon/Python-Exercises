```python
class Device:
    def __init__(self, name):
        self.name = name
        self.app_list = []

    def __str__(self):
        return f"Urządzenie{self.name}"

class Phone(Device):
    def __init__(self, name):
        super().__init__(name)

    def __str__(self):
        return f"Telefon{self.name}"

class Smartphone(Phone):
    def install_app(self, app_name):
        self.app_list.append(app_name)
        print(f"Aplikacja '{app_name}' została pomyślnie zainstalowana na {self.name}.")

    def __str__(self):
        apps = ", ".join(self.app_list) if self.app_list else "brak aplikacji"
        return f"Smartfon{self.name}, aplikacje: {apps}"

class Tablet(Device):
    def install_app(self, app_name):
        self.app_list.append(app_name)
        print(f"Aplikacja '{app_name}' została zainstalowana na {self.name}.")

    def __str__(self):
        apps = ", ".join(self.app_list) if self.app_list else "brak aplikacji"
        return f"Tablet{self.name}, aplikacje: {apps}"

class Computer(Device):
    def __str__(self):
        return f"Komputer{self.name}"

class Laptop(Computer):
    def __str__(self):
        return f"Laptop{self.name}"

class Smartwatch(Device):
    def __str__(self):
        return f"Smartwatch{self.name}"

s = Smartphone("Samsung Galaxy S25")
t = Tablet("Lenovo Tab M11")

s.install_app("YouTube")
s.install_app("Spotify")
t.install_app("Netflix")

print()
print(s)
print(t)

```
