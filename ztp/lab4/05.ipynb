Proszę utworzyć funkcję licznik(start), która zwraca funkcję wewnętrzną zwiększającą licznik przy każdym wywołaniu.

```
c = licznik(10)
print(c())  # 11
print(c())  # 12
```

````python
def licznik(start):
    def zwieksz():
        nonlocal start
        start = start + 1
        return start
    return zwieksz

c = licznik(10)
print(c())
print(c())
````
