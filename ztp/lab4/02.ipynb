Proszę utworzyć listę trzech wbudowanych funkcji, np. lista = [len, sum, max].
Proszę wypisać nazwę każdej funkcji (korzystając z atrybutu __name__.) oraz jej wynik

```
lista_funkcji = [len, sum, max]
lista_arg = list(range(10))
naszaf(lista_funkcji, lista_arg)

wynik działania:
'len', 5
'sum', 15
'max', 5
```


````python
def result(name, args):
    for x in name:
        print(f" '{x.__name__}', {x(args)}") 

f_list = [len, sum, max]
arg_list = list(range(10))

result(f_list, arg_list)
````
