Proszę przygotować dekorator @measure_time, który zmierzy i wypisze czas działania funkcji.

```
@measure_time
def slow_function():
    for _ in range(10**6):
        pass
```

````python
import time

def measure_time(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        res = func(*args, **kwargs)
        end = time.time()
        durr = end - start
        print(f"Czas działania funkcji {func.__name__}: {durr:.4f}s")
        return res
    return wrapper
    
@measure_time
def slow_function():
    for _ in range(10**6):
        pass
        
slow_function()
````
