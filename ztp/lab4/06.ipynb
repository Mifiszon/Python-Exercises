Proszę utworzyć dekorator @logger, który przed wywołaniem funkcji wypisze:
` [LOG] Wywołano funkcję <nazwa> `

```
@logger
def hello():
    print("Hej!")

[LOG] Wywołano funkcję hello
Hej!
```

````python
def logger(func):
    def wrapper():
        print(f"[LOG] Wywołano funkcję <{func.__name__}>")
        func()
    return wrapper

@logger
def hello():
    print("Hej!")

hello()
````
