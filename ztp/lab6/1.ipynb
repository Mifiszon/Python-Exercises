```python
import re
path = "./KORPUSY/POL/POL1 Pan Tadeusz.txt"

with open(path, 'r', encoding="utf-8") as file:
    text = file.read()
    text = text.lower()
    
    chars = ['.', ' \n', '\n ', "'", '"', '%', '_', ',', '-', '»', ';', '?', '!', '[', ']', '«', '(', ')', '  ', '   ', ':', '}', '\n', '    ']

    for c in chars:
        text = text.replace(c, " ")
    
    text = re.sub(r"\s+", " ", text).strip()

    words = re.findall('[a-ż]+', text)

    uni = {}
    for word in words:
        for letter in word:
            if letter in uni:
                uni[letter] += 1
            else:
                uni[letter] = 1

    suni = sorted(uni.items(), key=lambda x: x[1], reverse=True)

    #for letter, count in suni:
    #    print(f"{letter}: {count}")

    di = {}
    for i in range(len(text)-1):
        if text[i].isalpha() or text[i] == " ":
            if text[i]+text[i+1] in di:
                di[text[i]+text[i+1]] += 1
            else:
                di[text[i]+text[i+1]] = 1
            
    sdi = sorted(di.items(), key=lambda x: x[1], reverse=True)
    
    #for pair, count in sdi:
    #    print(f"{pair}: {count}")

    tri = {}
    for i in range(len(text)-2):
        if text[i].isalpha() or text[i] == " ":
            if text[i]+text[i+1]+text[i+2] in tri:
                tri[text[i]+text[i+1]+text[i+2]] += 1
            else:
                tri[text[i]+text[i+1]+text[i+2]] = 1

    stri = sorted(tri.items(), key=lambda x: x[1], reverse=True)
    
    #for triple, count in stri:
    #    print(f"{triple}: {count}")

    with open("unigrams.csv", "w", encoding="utf-8") as f:
        f.write("ngram,count\n")
        for letter, count in suni:
            f.write(f"{letter},{count}\n")
    
    with open("digrams.csv", "w", encoding="utf-8") as f:
        f.write("ngram,count\n")
        for pair, count in sdi:
            f.write(f"{pair},{count}\n")
    
    with open("trigrams.csv", "w", encoding="utf-8") as f:
        f.write("ngram,count\n")
        for triple, count in stri:
            f.write(f"{triple},{count}\n")
```
