**Wszystkie klasy** powinny umożliwiać porównywanie przy pomocy '==' i '!='. 

Dwa punkty są równe, gdy maja wszystkie współrzędne o równych wartościach. 

Dwa trójkąty są równe, gdy zawierają punkty parami równe, obojętnie w jakiej kolejności.

**Wszystkie metody** powinny być udokumentowane.

````python
import math

class Triangle:
    "Trójkąt w trójwymiarowej przestrzeni euklidesowej."
    def __init__(self, point1, point2, point3):
        "konstruktor klasy trinagle, który przyjmuje 3 punkty jako arguemnty"
        self.point1 = point1
        self.point2 = point2
        self.point3 = point3

    def __repr__(self):
        "reprezetnacja tekstowa trójkąta"
        return 'triangle: %s, %s, %s' % (self.point1, self.point2, self.point3)

    def perimeter(self):
        "Oblicza obwód trójkąta."
        vector1 = (self.point2 - self.point1).norm()
        vector2 = (self.point3 - self.point2).norm()
        vector3 = (self.point1 - self.point3).norm()
        return vector1 + vector2 + vector3

    def area(self):
        "Oblicza pole powierzchni trójkąta"
        vector1 = (self.point2 - self.point1).norm()
        vector2 = (self.point3 - self.point2).norm()
        vector3 = (self.point1 - self.point3).norm()
        halfp = 0.5 * (vector1 + vector2 + vector3)
        return math.sqrt(halfp * (halfp-vector1) * (halfp-vector2) * (halfp-vector3))

    def __eq__(self, tri):
        "sprawdza czy 2 trójkąty są równe"
        return (
            (self.point1 == tri.point1 and self.point2 == tri.point2 and self.point3 == tri.point3) or
            (self.point1 == tri.point1 and self.point2 == tri.point3 and self.point3 == tri.point2) or
            (self.point1 == tri.point2 and self.point2 == tri.point1 and self.point3 == tri.point3) or
            (self.point1 == tri.point2 and self.point2 == tri.point3 and self.point3 == tri.point1) or
            (self.point1 == tri.point3 and self.point2 == tri.point1 and self.point3 == tri.point2) or
            (self.point1 == tri.point3 and self.point2 == tri.point2 and self.point3 == tri.point1)
        )

    def __ne__(self, tri):
        "sprawdza czy 2 trójkąty nie są równe"
        return not self == tri

class Vector:
    "Wektor w trójwymiarowej przestrzeni euklidesowej."
    def __init__(self, x, y, z):
        "konstruktor klasy vector"
        self.x = float(x)
        self.y = float(y)
        self.z = float(z)

    def __repr__(self):
        "reprezentacja tekstowa vectora"
        return '[%.2f, %.2f, %.2f]' % (self.x, self.y, self.z)

    def __add__(self, vec):
        "suma: self + vec - dodanie 2 vectorów"
        return Vector(self.x + vec.x, self.y + vec.y, self.z + vec.z)

    def __eq__(self, vec):
        "sprawdza czy 2 vectory są równe"
        return self.x == vec.x and self.y == vec.y and self.z == vec.z

    def __ne__(self, vec):
        "sprawdza czy 2 vectory nie są równe"
        return not self == vec

    def __mul__(self, vec):
        "iloczyn skalarny 2 vectorów"
        return self.x * vec.x + self.y * vec.y + self.z * vec.z

    def norm(self):
        "oblicza długość vectora"
        return math.sqrt((self.x ** 2)+(self.y ** 2)+(self.z ** 2))

    def _bool_(self):
        "sprawdza czy vector istnieje"
        if (self.x == 0 and self.y == 0 and self.z == 0):
            return False
        else:
            return True

class Point:
    "Punkt w trójwymiarowej przestrzeni euklidesowej."
    def __init__(self, x, y, z):
        "konstruktor klasy point, współrzędne punktu jako argumenty"
        self.x = round(float(x), 2)
        self.y = round(float(y), 2)
        self.z = round(float(z), 2)

    def __repr__(self):
        "reprezetnacja tekstowa klasy point"
        return '(%.2f, %.2f, %.2f)' % (self.x, self.y, self.z)

    def __sub__(self, pt):
        "obliczanie vectora, czyli odległości punktó od siebie)
        return Vector(self.x - pt.x, self.y - pt.y, self.z - pt.z)

    def __add__(self, vec):
        "dodanie vectora do punktu i tworzy nowy"
        return Point(self.x + vec.x, self.y + vec.y, self.z + vec.z)

    def __iadd__(self, vec):
        "suma: self += vec - dodanie vector do point za pomoca iteratora"
        self.x += vec.x
        self.y += vec.y
        self.z += vec.z
        return self

    def __eq__(self, pt):
        "sprwadza czy 2 punkty sa równe"
        return self.x == pt.x and self.y == pt.y and self.z == pt.z

    def __ne__(self, pt):
        "sprawdza czy 2 punkty sa nierówne"
        return not self == pt

A = Point(0, 0, 0)
B = Point(1, 0, 0)
C = Point(0, 1, 0)
t = Triangle(A, B, C)
print(t)
print(t.perimeter())
print(t.area())
````
